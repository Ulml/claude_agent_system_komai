```python
# Création d'un graphe orienté pour l'assemblage des composants
G_assemblage = nx.DiGraph()

# Ajout des nœuds pour les composants
composants = [
    "Béton armé", "Acier", "Gravier", "Blocs de terre", "Mortier de terre",
    "Enduit terre", "Bois", "Tuiles/tôle", "Isolant", "Portes",
    "Fenêtres", "Dalle terre", "Revêtement", "Câbles", "Tuyaux", "Peinture"
]

# Ajout des nœuds pour les étapes d'assemblage
etapes_assemblage = [
    "Fondations", "Murs en terre", "Charpente/Toiture",
    "Menuiseries", "Sols", "Électricité/Plomberie", "Finition intérieure"
]

# Ajout des nœuds pour les sous-ensembles
sous_ensembles = [
    "Structure de base", "Superstructure", "Intérieur", "Maison en terre 200 m²"
]

# Ajout de tous les nœuds
G_assemblage.add_nodes_from(composants, type="composant")
G_assemblage.add_nodes_from(etapes_assemblage, type="étape")
G_assemblage.add_nodes_from(sous_ensembles, type="sous-ensemble")

# Ajout des arêtes pour l'assemblage des composants aux étapes
arêtes_composants_étapes = [
    ("Béton armé", "Fondations"), ("Acier", "Fondations"), ("Gravier", "Fondations"),
    ("Blocs de terre", "Murs en terre"), ("Mortier de terre", "Murs en terre"),
    ("Enduit terre", "Murs en terre"),
    ("Bois", "Charpente/Toiture"), ("Tuiles/tôle", "Charpente/Toiture"),
    ("Isolant", "Charpente/Toiture"),
    ("Portes", "Menuiseries"), ("Fenêtres", "Menuiseries"),
    ("Dalle terre", "Sols"), ("Revêtement", "Sols"),
    ("Câbles", "Électricité/Plomberie"), ("Tuyaux", "Électricité/Plomberie"),
    ("Enduit terre", "Finition intérieure"), ("Peinture", "Finition intérieure")
]

# Ajout des arêtes pour les étapes aux sous-ensembles
arêtes_étapes_sous_ensembles = [
    ("Fondations", "Structure de base"),
    ("Murs en terre", "Superstructure"),
    ("Charpente/Toiture", "Superstructure"),
    ("Menuiseries", "Intérieur"),
    ("Sols", "Intérieur"),
    ("Électricité/Plomberie", "Intérieur"),
    ("Finition intérieure", "Intérieur")
]

# Ajout des arêtes pour les sous-ensembles au produit final
arêtes_sous_ensembles_produit = [
    ("Structure de base", "Maison en terre 200 m²"),
    ("Superstructure", "Maison en terre 200 m²"),
    ("Intérieur", "Maison en terre 200 m²")
]

# Ajout de toutes les arêtes
G_assemblage.add_edges_from(arêtes_composants_étapes)
G_assemblage.add_edges_from(arêtes_étapes_sous_ensembles)
G_assemblage.add_edges_from(arêtes_sous_ensembles_produit)

# Dessin du graphe
plt.figure(figsize=(16, 10))
pos_assemblage = nx.spring_layout(G_assemblage, seed=42, k=0.5)

# Dessiner les nœuds avec des couleurs différentes selon leur type
node_colors = []
for node in G_assemblage.nodes():
    if G_assemblage.nodes[node]['type'] == 'composant':
        node_colors.append('lightgreen')
    elif G_assemblage.nodes[node]['type'] == 'étape':
        node_colors.append('skyblue')
    else:
        node_colors.append('lightcoral')

nx.draw(
    G_assemblage, pos_assemblage, with_labels=True,
    node_size=3000, node_color=node_colors,
    font_size=10, font_weight='bold', arrowsize=20
)

# Légende
plt.scatter([], [], c='lightgreen', label='Composant')
plt.scatter([], [], c='skyblue', label='Étape')
plt.scatter([], [], c='lightcoral', label='Sous-ensemble/Produit final')
plt.legend(scatterpoints=1, frameon=False, labelspacing=1, loc='upper right')

# Titre et affichage
plt.title("Graphe d'assemblage des composants pour une maison en terre 200 m²")
plt.show()
```
